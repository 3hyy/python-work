## 欧几里得算法

### 1. 实现 `gcd` 函数

````python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m
````

### 2. 计算给定数对的 GCD

````python
print("gcd(2018, 2019) =", gcd(2018, 2019))
print("gcd(1200, 300)  =", gcd(1200, 300))
print("gcd(5040, 60)   =", gcd(5040, 60))
````

### 3. 负数参数的行为

用负数参数测试：

````python
print("gcd(-2018, 2019) =", gcd(-2018, 2019))
print("gcd(1200, -300)  =", gcd(1200, -300))
print("gcd(-1200, -300) =", gcd(-1200, -300))
````

**结论：** 当参数含负数时，函数依赖 Python 取模运算（`%` 的结果符号跟随除数），结果的符号不确定：`gcd(-2018, 2019)` 返回正的 `1`，但当第二个参数为负、且两数恰好整除时（如 `gcd(1200, -300)` 和 `gcd(-1200, -300)`），会返回负数 `-300`。

这种行为**没有意义**，因为按定义 GCD 是同时整除两个数的**最大正整数**，恒为非负。修复方法是在函数开头对两个参数取绝对值：

```python
def gcd(m, n):
    m, n = abs(m), abs(n)
    while n != 0:
        m, n = n, m % n
    return m
```

## 近似欧拉数 e

### 1. `euler_limit`：利用极限公式 $(1+\frac{1}{n})^n$

````python
def euler_limit(n):
    return (1 + 1 / n) ** n

print(euler_limit(10))
print(euler_limit(100000))
````

### 2. `euler_infinite_sum`：利用公式 (1) 的前 `n` 项之和

注意函数是“1-索引”的：`euler_infinite_sum(0)` 返回 `0`，`euler_infinite_sum(1)` 只含 $k=0$ 这一项，返回 `1`。

````python
def euler_infinite_sum(n):
    total = 0.0
    factorial = 1          # 0! = 1
    for k in range(n):     # 累加 k = 0, 1, ..., n-1 共 n 项
        if k > 0:
            factorial *= k
        total += 1 / factorial
    return total

print(euler_infinite_sum(0))
print(euler_infinite_sum(1))
print(euler_infinite_sum(4))
````

### 3. `euler_approx`：不断累加，直到近似值与真实 `e` 的误差小于 `epsilon`

````python
import math

def euler_approx(epsilon):
    n = 0
    while math.exp(1) - euler_infinite_sum(n) >= epsilon:
        n += 1
    return euler_infinite_sum(n)

approx = euler_approx(1e-6)
print("近似值:", approx)
print("真实值:", math.exp(1))
print("误差:  ", math.exp(1) - approx)
````

### 4. `print_euler_sum_table`：逐行打印 `euler_infinite_sum(k)`，`k` 从 1 到 `n`

````python
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))

print_euler_sum_table(10)
````

## 测试整数属性

### 1. `is_power_of_2`：判断正整数是否为 2 的幂（只使用除法和取模）

````python
def is_power_of_2(n):
    while n % 2 == 0:
        n = n // 2
    return n == 1

for n in [1, 2, 4, 8, 16, 3, 10, 1024, 1023]:
    print(n, is_power_of_2(n))
````

### 2. `is_power(b, n)`：推广为判断 `n` 是否为 `b` 的幂

注意边界情况：$b^0 = 1$，所以任意 `b` 下 `n = 1` 都为 `True`；`b = 1` 时只有 `n = 1` 为 `True`。

````python
def is_power(b, n):
    if n == 1:
        return True
    if b == 1:
        return False
    while n % b == 0:
        n = n // b
    return n == 1

print(is_power(2, 8))     # True, 8 = 2^3
print(is_power(3, 9))     # True, 9 = 3^2
print(is_power(2, 10))    # False
print(is_power(5, 625))   # True, 625 = 5^4
print(is_power(1, 1))     # True
print(is_power(1, 5))     # False
````
